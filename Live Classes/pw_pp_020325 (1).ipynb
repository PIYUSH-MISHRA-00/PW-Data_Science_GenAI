```python
class Car:
    def __init__(self, brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color
        
    def display_info(self):
        return f"Car: {self.brand}{self.model}{self.color}"

car1 = Car("Toyota","Corolla","Black")
print(car1.display_info())
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
        
    def area(self):
        return self.length*self.width
    
    def perimeter(self):
        return 2*(self.length + self.width)
    
rect = Rectangle(5,4)

print(rect.area())
print(rect.perimeter())
```

```python
class Employee:
    def __init__(self, name, salary = 50000):
        self.name = name
        self.salary = salary
        
    def display_info(self):
        return f"Employee: {self.name}, Salary: {self.salary}"
    
e1 = Employee("Alice")
e2 = Employee("Bob")

print(e1.display_info())
print(e2.display_info())
```

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.__age = age #private attribute
        
    def get_age(self):
        return self.__age #getter
    
    def set_age(self, new_age):
        if new_age > 0:
            self.__age = new_age
        else:
            print("age cant be negative")
            
s1 = Student("Alice", 20)
print(s1.get_age()) 

s1.set_age(25)
print(s1.get_age())

s1.set_age(-5)
```

1. Single inheritance - one parent, one child
2. multiple inheritance - a child inherits from multiple parents
3. multilevel - a child inherits from a parent, which itself is a child of another class
4. hierarchical  -  multiple child classes inherit from same parent

```python
class Animal: #parent class
    def speak(self):
        return "animal makes a sound"
    
class Dog(Animal):
    def bark(self):
        return "dog barks"
    
d = Dog()
print(d.speak())
print(d.bark())
```

```python
class Engine:
    def start(self):
        return "Engine starts"
    
class Wheels:
    def rotate(self):
        return "Wheels rotating"
    
class Car(Engine, Wheels):
    def drive(self):
        return "Car is driving"
    
c = Car()
print(c.start())
print(c.rotate())
print(c.drive())
```

```python
class Animal:
    def live(self):
        return "animals are alive"
    
class Mammal(Animal):
    def feed(self):
        return "Mammals feed their young ones"
    
class Dog(Mammal):
    def bark(self):
        return "dog barks"
    
d = Dog()
print(d.live())
print(d.feed())
print(d.bark())
```

```python
class Animal:
    def speak(self):
        return "animal makes a sound"
    
class Dog(Animal):
    def bark(self):
        return "Dog barks"
    
class Cat(Animal):
    def meow(self):
        return "Cat meows"
    
c = Cat()
d = Dog()

print(c.speak())
print(d.speak())
print(c.bark())
print(d.meow())
```

```python
class Animal:
    def speak(self):
        return "animal makes a sound"
    
class Dog(Animal):
    def speak(self):
        return "Dog barks"
    
class Cat(Animal):
    def speak(self):
        return "Cat meows"
    
c = Cat()
d = Dog()
a = Animal()

print(c.speak())
print(d.speak())
print(a.speak())
print(c.a.speak())
```

Polymorphism

```python
class Animal:
    def speak(self):
        return "animal makes a sound"
    
class Dog(Animal):
    def speak(self): #method overriding
        return "Dog barks"
    
class Cat(Animal):
    def speak(self): #method overriding
        return "Cat meows"
    
animals = [Animal(), Cat(), Dog()]

for animal in animals:
    print(animal.speak())
```

```python
class Mo:
    def add(self, a, b=0, c=0):
        return a + b + c
    
m = Mo()
print(m.add(5))
print(m.add(5,10))
print(m.add(5,10,15))
```
