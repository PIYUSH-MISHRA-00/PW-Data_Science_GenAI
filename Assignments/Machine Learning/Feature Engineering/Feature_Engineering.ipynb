### 1. What is a parameter?
A parameter is a variable in a model or function that is learned from data during the training process. In machine learning, parameters are the internal coefficients or weights that the algorithm adjusts to minimize the error and best fit the data.

### 2. What is correlation?
Correlation is a statistical measure that describes the strength and direction of a relationship between two variables. It is commonly measured using Pearson's correlation coefficient, which ranges from -1 to 1.

### 3. What does negative correlation mean?
Negative correlation means that as one variable increases, the other variable tends to decrease. The correlation coefficient for a perfect negative correlation is -1.

### 4. Define Machine Learning. What are the main components in Machine Learning?
Machine Learning is a field of artificial intelligence that enables systems to learn from data and improve their performance without being explicitly programmed. The main components are:
- Data
- Model
- Loss function
- Optimizer
- Evaluation metric

### 5. How does loss value help in determining whether the model is good or not?
The loss value quantifies the difference between the predicted values and the actual values. A lower loss value indicates that the model's predictions are closer to the true values, suggesting a better model fit.

### 6. What are continuous and categorical variables?
Continuous variables can take any numerical value within a range (e.g., height, weight). Categorical variables represent discrete categories or groups (e.g., gender, color).

### 7. How do we handle categorical variables in Machine Learning? What are the common techniques?
Categorical variables are handled by encoding them into numerical values. Common techniques include Label Encoding, One-Hot Encoding, and Ordinal Encoding.

### 8. What do you mean by training and testing a dataset?
Training a dataset means using it to fit a model, while testing a dataset means evaluating the model's performance on unseen data to check its generalization ability.

### 9. What is sklearn.preprocessing?
`sklearn.preprocessing` is a module in scikit-learn that provides functions and classes for scaling, transforming, and encoding features to prepare data for machine learning models.

### 10. What is a Test set?
A test set is a portion of the dataset that is kept aside and used only to evaluate the performance of a trained machine learning model.

### 11. How do we split data for model fitting (training and testing) in Python?
We use the `train_test_split` function from `sklearn.model_selection` to split data into training and testing sets.

### 12. How do you approach a Machine Learning problem?
A typical approach includes: understanding the problem, collecting and cleaning data, performing EDA, feature engineering, model selection, training, evaluation, and deployment.

### 13. Why do we have to perform EDA before fitting a model to the data?
Exploratory Data Analysis (EDA) helps us understand the data, detect anomalies, identify patterns, and decide on feature engineering steps before modeling.

### 14. How can you find correlation between variables in Python?
You can use the `.corr()` method in pandas to compute the correlation matrix for a DataFrame. Example: `df.corr()`.

### 15. What is causation? Explain difference between correlation and causation with an example.
Causation means that one variable directly affects another. Correlation only shows a relationship, not cause. Example: Ice cream sales and drowning rates are correlated (both rise in summer), but eating ice cream does not cause drowning.

### 16. What is an Optimizer? What are different types of optimizers? Explain each with an example.
An optimizer is an algorithm that updates model parameters to minimize the loss function. Common optimizers:
- Gradient Descent: Updates parameters in the direction of the negative gradient.
- Stochastic Gradient Descent (SGD): Uses random samples for updates.
- Adam: Combines momentum and adaptive learning rates.
Example: In TensorFlow/Keras, you can use `optimizer = tf.keras.optimizers.Adam()`.

### 17. What is sklearn.linear_model?
`sklearn.linear_model` is a module in scikit-learn that provides linear models for regression and classification, such as LinearRegression and LogisticRegression.

### 18. What does model.fit() do? What arguments must be given?
`model.fit()` trains the model on the training data. It requires at least the feature matrix `X` and the target vector `y` as arguments.

### 19. What does model.predict() do? What arguments must be given?
`model.predict()` uses the trained model to make predictions on new data. It requires the feature matrix `X` as an argument.

Feature scaling is the process of normalizing or standardizing the range of independent variables. It helps algorithms converge faster and ensures that features contribute equally to the result.

### 22. Explain data encoding.
Data encoding is the process of converting categorical data into numerical format so that machine learning algorithms can use it. Common methods include label encoding and one-hot encoding.
