## Theoretical Questions

1. What is the difference between interpreted and compiled languages?  
Ans. Interpreted languages execute code line by line at runtime, while compiled languages are translated into machine code before execution.

2. What is exception handling in Python?  
Ans. Exception handling in Python is the process of responding to runtime errors using try, except, else, and finally blocks.

3. What is the purpose of the `finally` block in exception handling?  
Ans. The `finally` block ensures that specific cleanup code executes regardless of whether an exception occurred.

4. What is logging in Python?  
Ans. Logging in Python is the practice of recording events, errors, and other runtime information to help in debugging and monitoring applications.

5. What is the significance of the `__del__` method in Python?  
Ans. The `__del__` method acts as a destructor, called when an object is about to be destroyed, allowing for resource cleanup.

6. What is the difference between `import` and `from ... import` in Python?  
Ans. `import module` brings in the entire module and its namespace, while `from module import name` imports specific attributes directly into the current namespace.

7. How can you handle multiple exceptions in Python?  
Ans. By using multiple except clauses or by specifying a tuple of exception types in a single except clause.

8. What is the purpose of the `with` statement when handling files in Python?  
Ans. The `with` statement simplifies file handling by ensuring that the file is automatically closed after its block of code is executed.

9. What is the difference between multithreading and multiprocessing?  
Ans. Multithreading runs multiple threads within a single process sharing the same memory, whereas multiprocessing runs separate processes, each with its own memory space.

10. What are the advantages of using logging in a program?  
Ans. Logging helps in debugging, monitoring application behavior, and provides a persistent record of events which is useful for troubleshooting.

11. What is memory management in Python?  
Ans. Memory management in Python refers to how the interpreter allocates and recycles memory for objects, primarily handled through reference counting and garbage collection.

12. What are the basic steps involved in exception handling in Python?  
Ans. The steps include encapsulating risky code in a try block, catching exceptions with except blocks, optionally executing code with an else block when no errors occur, and cleaning up with a finally block.

13. Why is memory management important in Python?  
Ans. It ensures efficient use of resources, prevents memory leaks, and maintains the performance and stability of applications.

14. What is the role of `try` and `except` in exception handling?  
Ans. The try block contains code that might raise an exception, and the except block handles the exception if one occurs, preventing program crashes.

15. How does Python's garbage collection system work?  
Ans. Python primarily uses reference counting, supplemented by a cyclic garbage collector to reclaim memory from objects involved in circular references.

16. What is the purpose of the `else` block in exception handling?  
Ans. The `else` block runs if the try block completes without raising any exceptions, useful for code that should run only when no errors occur.

17. What are the common logging levels in Python?  
Ans. Common logging levels include DEBUG, INFO, WARNING, ERROR, and CRITICAL.

18. What is the difference between `os.fork()` and multiprocessing in Python?  
Ans. `os.fork()` creates a new process by duplicating the current process (available on Unix), whereas the multiprocessing module offers a platform-independent interface to create and manage processes.

19. What is the importance of closing a file in Python?  
Ans. Closing a file releases system resources and ensures that all buffered data is properly written to disk.

20. What is the difference between `file.read()` and `file.readline()` in Python?  
Ans. `file.read()` reads the entire content of a file at once, while `file.readline()` reads the file one line at a time.

21. What is the logging module in Python used for?  
Ans. The logging module is used to record messages that describe events during program execution for debugging and monitoring purposes.

22. What is the `os` module in Python used for in file handling?  
Ans. The `os` module provides functions for interacting with the operating system, including file and directory manipulation.

23. What are the challenges associated with memory management in Python?  
Ans. Challenges include dealing with circular references, optimizing garbage collection, and balancing performance with efficient memory usage.

24. How do you raise an exception manually in Python?  
Ans. By using the raise statement, e.g., raise Exception("Error message").

25. Why is it important to use multithreading in certain applications?  
Ans. Multithreading can improve performance and responsiveness in I/O-bound or concurrent tasks by allowing multiple threads to run simultaneously.

```python
# 1. How can you open a file for writing in Python and write a string to it?

with open("output.txt", "w") as file:
    file.write("This is a sample string.")
```

```python
# 2. Write a Python program to read the contents of a file and print each line.

with open("output.txt", "r") as f:
    for line in f:
        print(line.rstrip())
```

```python
# 3. How would you handle a case where the file doesn't exist while trying to open it for reading?

try:
    with open("nonexistent.txt", "r") as f:
        content = f.read()
        print(content)
except FileNotFoundError:
    print("The file 'nonexistent.txt' does not exist.")
```

```python
# 4. Write a Python script that reads from one file and writes its content to another file.

with open("output.txt", "r") as read_file:
    content = read_file.read()
with open("output_copy.txt", "w") as out_file:
    out_file.write(content)
```

```python
# 5. How would you catch and handle a division by zero error in Python?

try:
    result = 10 / 0
except ZeroDivisionError:
    print("Caught a division by zero error!")
```

```python
# 6. Write a Python program that logs an error message to a log file when a division by zero exception occurs.

import logging
logging.basicConfig(filename='error.log', level=logging.ERROR, format='%(asctime)s:%(levelname)s:%(message)s')

try:
    result = 10 / 0
except ZeroDivisionError:
    logging.error("Division by zero error occurred", exc_info=True)
```

```python
# 7. How do you log information at different levels (`INFO`, `ERROR`, `WARNING`) in Python using the logging module?

logging.info("This is an INFO level message.")
logging.warning("This is a WARNING level message.")
logging.error("This is an ERROR level message.")
```

```python
# 8. Write a program to handle a file opening error using exception handling.

try:
    with open("nonexistent_file.txt", "r") as file:
        data = file.read()
except FileNotFoundError as e:
    print("Error: File not found.", e)
```

```python
# 9. How can you read a file line by line and store its content in a list in Python?

with open("output.txt", "r") as read_file:
	lines_list = read_file.readlines()
print(lines_list)
```

```python
# 10. How can you append data to an existing file in Python?

with open("output.txt", "a") as f:
    f.write("\nAppended data.")
```

```python
# 11. Write a Python program that uses a `try-except` block to handle an error when attempting to access a dictionary key that doesn't exist.

sample_dict = {"a": 1, "b": 2}
try:
    value = sample_dict["c"]
except KeyError:
    print("The key 'c' does not exist in the dictionary.")
```

```python
# 12. Write a program that demonstrates using multiple `except` blocks to handle different types of exceptions.

try:
    # This will raise a KeyError because key 'c' is not in sample_dict
    print(sample_dict["c"])
    # This line would raise a ZeroDivisionError if reached
    result = 10 / 0
except ZeroDivisionError:
    print("Caught a division by zero error!")
except KeyError:
    print("Caught a key error!")
```

```python
# 13. How would you check if a file exists before attempting to read it in Python?

import os
if os.path.exists("output.txt"):
    with open("output.txt", "r") as file:
        data = file.read()
        print(data)
else:
    print("File does not exist.")
```

```python
# 14. Write a program that uses the logging module to log both informational and error messages.

logging.info("This is an informational message.")
logging.error("This is an error message.")
```

```python
# 15. Write a Python program that prints the content of a file and handles the case when the file is empty.

with open("output.txt", "r") as file:
    contents = file.read()
if contents:
    print(contents)
else:
    print("The file is empty.")
```

```python
# 16. Demonstrate how to use memory profiling to check the memory usage of a small program.

from memory_profiler import memory_usage
def some_function():
    # Simulate memory usage by creating a large list
    large_list = [i for i in range(100000)]
    return sum(large_list)

usage = memory_usage(some_function, interval=0.1)
print("Memory usage (MB):", usage)
```

```python
# 17. Write a Python program to create and write a list of numbers to a file, one number per line.

numbers = [i for i in range(1, 11)]
with open("numbers.txt", "w") as file_numbers:
    for number in numbers:
        file_numbers.write(f"{number}\n")
```

```python
# 18. How would you implement a basic logging setup that logs to a file with rotation after 1MB?

from logging.handlers import RotatingFileHandler
logger = logging.getLogger(__name__)
logger.setLevel(logging.INFO)

# Create a RotatingFileHandler that logs to 'app.log' with a max size of 1MB and keeps 3 backup files.
handler = RotatingFileHandler("app.log", maxBytes=1*1024*1024, backupCount=3)
formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
handler.setFormatter(formatter)

logger.addHandler(handler)

logger.info("This is an info message to demonstrate rotating file handler.")
```

```python
# 19. Write a program that handles both `IndexError` and `KeyError` using a `try-except` block.

try:
    # This will raise an IndexError since lines_list has only one element.
    print(lines_list[1])
    # This will raise a KeyError since key 'c' does not exist in sample_dict.
    print(sample_dict["c"])
except IndexError:
    print("Caught an IndexError!")
except KeyError:
    print("Caught a KeyError!")
```

```python
# 20. How would you open a file and read its contents using a context manager in Python?

with open("output.txt", "r") as file:
    file_contents = file.read()
print(file_contents)
```

```python
# 21. Write a Python program that reads a file and prints the number of occurrences of a specific word.

with open("output.txt", "r") as file:
    text = file.read()
word = "sample"  # you can change this to any word you want to count
occurrences = text.count(word)
print(f"The word '{word}' appears {occurrences} times.")
```

```python
# 22. How can you check if a file is empty before attempting to read its contents?

if os.path.getsize("output.txt") > 0:
    with open("output.txt", "r") as file_check:
        print(file_check.read())
else:
    print("The file is empty.")
```

```python
# 23. Write a Python program that writes to a log file when an error occurs during file handling.

import logging
logging.basicConfig(filename='file_error.log', level=logging.ERROR,
                    format='%(asctime)s:%(levelname)s:%(message)s')

try:
    # Attempt to open a file that does not exist to trigger an error
    with open("nonexistent_file.txt", "r") as f:
        f.read()
except Exception as e:
    logging.error("An error occurred during file handling: %s", e, exc_info=True)
```
