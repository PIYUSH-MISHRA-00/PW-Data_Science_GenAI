## Theoretical Questions

1. What is a RESTful API?

Ans. A RESTful API is an architectural style for designing networked applications. It is based on the idea of resources, which are identified by URIs, and can be manipulated using a fixed set of operations. RESTful APIs are typically stateless, cacheable, and use HTTP methods like GET, POST, PUT, and DELETE to interact with resources. They are often used for building web services and microservices.

2. Explain the concept of API specification.

Ans. An API specification outlines how an API should work by describing its endpoints, request and response formats, authentication methods, and error handling. It serves as a contract to ensure consistency between different parts of the system and across teams.

3. What is Flask, and why is it popular for building APIs?

Ans. Flask is a lightweight Python web framework that is easy to set up and flexible. It is popular for building APIs due to its simplicity, extensive documentation, and a large ecosystem of extensions that support various functionalities.

4. What is routing in Flask?

Ans. Routing in Flask refers to the mechanism of mapping URL patterns to Python functions (view functions). This determines how Flask handles different URL requests and directs them to the corresponding function for processing.

5. How do you create a simple Flask application?

Ans. To create a simple Flask application, instantiate the Flask class, define routes using the @app.route() decorator, implement view functions that process requests and return responses, and then run the app using app.run().

6. What are HTTP methods used in RESTful APIs?

Ans. HTTP methods used in RESTful APIs include GET, POST, PUT, DELETE, PATCH, OPTIONS, and HEAD. Each method corresponds to a specific type of operation on the resource.

7. What is the purpose of the `@app.route()` decorator in Flask?

Ans. The `@app.route()` decorator in Flask binds a URL to a view function. When the specified URL is requested, the decorated function is executed to handle the request.

8. What is the difference between GET and POST HTTP methods?

Ans. GET is used to retrieve data from a server without altering its state, while POST is used to send data to the server to create or update resources. GET parameters are sent in the URL, whereas POST data is sent in the request body.

9. How do you handle errors in Flask APIs?

Ans. Errors in Flask APIs are handled by defining error handlers using the @app.errorhandler() decorator. This allows you to return custom responses for various HTTP error codes, improving the API’s robustness.

10. How do you connect Flask to a SQL database?

Ans. To connect Flask to a SQL database, you can use SQLAlchemy or Flask-SQLAlchemy. These libraries provide ORM capabilities, allowing you to define models, interact with the database, and perform CRUD operations in a Pythonic way.

11. What is the role of Flask-SQLAlchemy?

Ans. Flask-SQLAlchemy integrates the SQLAlchemy ORM with Flask, providing a simplified interface for database operations, model management, and query building while reducing boilerplate code.

12. What are Flask blueprints, and how are they useful?

Ans. Flask blueprints allow you to organize your application into modular components. They make it easier to manage, scale, and reuse code across different parts of the application, especially in larger projects.

13. What is the purpose of Flask's `request` object?

Ans. Flask's `request` object encapsulates all the details of the incoming HTTP request, including form data, query parameters, headers, and files. It allows you to access and process client-sent information within your view functions.

14. How do you create a RESTful API endpoint using Flask?

Ans. Create a RESTful API endpoint in Flask by defining a route with @app.route() that accepts specific HTTP methods, processing incoming data through the request object, and returning a response, typically formatted in JSON.

15. What is the purpose of Flask's `jsonify()` function?

Ans. Flask's `jsonify()` function converts Python dictionaries or lists into a JSON response with the appropriate MIME type, making it easier to return structured data from API endpoints.

16. Explain Flask’s `url_for()` function.

Ans. Flask’s `url_for()` function dynamically generates URLs for view functions based on their name and parameters, which helps in maintaining links even if URL patterns change over time.

17. How does Flask handle static files (CSS, JavaScript, etc.)?

Ans. Flask serves static files by default from the 'static' directory. These files are accessible via the /static/ URL path or by using url_for('static', filename='filename') in templates and view functions.

18. What is an API specification, and how does it help in building a Flask API?

Ans. An API specification defines the endpoints, methods, data formats, and other aspects of an API. It aids in the development process by providing clear guidelines, ensuring consistency, and facilitating communication between API providers and consumers.

19. What are HTTP status codes, and why are they important in a Flask API?

Ans. HTTP status codes indicate the result of an HTTP request (e.g., 200 for success, 404 for not found, 500 for server errors). They are important in a Flask API to inform clients about the outcome of their requests and to facilitate proper error handling.

20. How do you handle POST requests in Flask?

Ans. Handle POST requests in Flask by defining a route that accepts the POST method using methods=['POST'] in the @app.route() decorator, then accessing submitted data via the request object to process and respond accordingly.

21. How would you secure a Flask API?

Ans. Secure a Flask API by implementing authentication and authorization mechanisms, using HTTPS to encrypt data, validating and sanitizing inputs, and utilizing security extensions like Flask-JWT, Flask-Login, or implementing OAuth strategies.

22. What is the significance of the Flask-RESTful extension?

Ans. The Flask-RESTful extension simplifies the creation of RESTful APIs by providing higher-level abstractions for routing, request parsing, and response formatting, which helps in building clean and maintainable API code.

23. What is the role of Flask’s `session` object?

Ans. Flask’s `session` object is used to store and retrieve data specific to a user session. It allows you to persist user-related information between requests in a secure manner using client-side cookies.

```python
# 1. How do you create a basic Flask application?

from flask import Flask
app = Flask(__name__)

@app.route('/')
def home():
    return "Hello, Flask!"

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 2. How do you serve static files like images or CSS in Flask?

from flask import Flask, send_from_directory

app = Flask(__name__)

@app.route('/static/<path:path>')
def send_static(path):
    return send_from_directory('static', path)

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 3. How do you define different routes with different HTTP methods in Flask?

from flask import Flask, jsonify, request

app = Flask(__name__)

@app.route('/users', methods=['GET'])
def get_users():
    users = [
        {'id': 1, 'name': 'John Doe'},
        {'id': 2, 'name': 'Jane Doe'}
    ]
    return jsonify(users)

@app.route('/users', methods=['POST'])
def create_user():
    data = request.json
    user = {'id': 3, 'name': data['name']}
    return jsonify(user), 201

@app.route('/users/<int:user_id>', methods=['GET'])
def get_user(user_id):
    user = {'id': user_id, 'name': 'John Doe'}
    return jsonify(user)

@app.route('/users/<int:user_id>', methods=['PUT'])
def update_user(user_id):
    data = request.json
    user = {'id': user_id, 'name': data['name']}
    return jsonify(user)

@app.route('/users/<int:user_id>', methods=['DELETE'])
def delete_user(user_id):
    return jsonify({'message': 'User deleted'}), 200
```

```python
# 4. How do you render HTML templates in Flask?
from flask import Flask, render_template

app = Flask(__name__)

@app.route('/')
def index():
    return render_template('index.html')

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 5. How can you generate URLs for routes in Flask using `url_for`?
from flask import Flask, url_for

app = Flask(__name__)

@app.route('/')
def home():
    return 'Home Page'

@app.route('/hello')
def hello():
    return 'Hello World'

@app.route('/user/<username>')
def show_user_profile(username):
    return 'User %s' % username

@app.route('/post/<int:post_id>')
def show_post(post_id):
    return 'Post %d' % post_id

@app.route('/path/<path:subpath>')
def show_subpath(subpath):
    return 'Subpath %s' % subpath

@app.route('/login')
def login():
    return 'Login Page'

@app.route('/user/<username>/followed')
def show_followed_by(username):
    return 'Followed by %s' % username

@app.route('/user/<username>/followers')
def show_followers_of(username):
    return 'Followers of %s' % username

with app.test_request_context('/hello'):
    print(url_for('static', filename='style.css'))
    print(url_for('home'))
    print(url_for('hello'))
    print(url_for('show_user_profile', username='John Doe'))
    print(url_for('show_post', post_id=1))
    print(url_for('show_subpath', subpath='/path/to/subpath'))
    print(url_for('login'))
    print(url_for('show_followed_by', username='John Doe'))
    print(url_for('show_followers_of', username='John Doe'))
```

```python
# 6. How do you handle forms in Flask?

from flask import Flask, request, render_template_string
from flask_wtf import FlaskForm
from wtforms import StringField, SubmitField
from wtforms.validators import DataRequired

app = Flask(__name__)
app.config['SECRET_KEY'] = 'mysecretkey'

class MyForm(FlaskForm):
    name = StringField('Name', validators=[DataRequired()])
    submit = SubmitField('Submit')

@app.route('/', methods=['GET', 'POST'])
def index():
    form = MyForm()
    if form.validate_on_submit():
        name = form.name.data
        return f'Hello, {name}!'
    return render_template_string('''
    <form method="POST">
        {{ form.hidden_tag() }}
        {{ form.name.label }} {{ form.name() }}
        {{ form.submit() }}
    </form>
    ''', form=form)

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 7. How can you validate form data in Flask?

from flask import Flask, request, jsonify
from flask_wtf import FlaskForm
from wtforms import StringField, IntegerField
from wtforms.validators import DataRequired

app = Flask(__name__)
app.config['SECRET_KEY'] = 'mysecretkey'

class MyForm(FlaskForm):
    name = StringField('name', validators=[DataRequired()])
    age = IntegerField('age', validators=[DataRequired()])

@app.route('/submit', methods=['POST'])
def submit():
    form = MyForm()
    if form.validate_on_submit():
        name = form.name.data
        age = form.age.data
        return jsonify({'message': 'Form data is valid', 'name': name, 'age': age})
    else:
        return jsonify({'message': 'Form data is invalid', 'errors': form.errors}), 400
```

```python
# 8. How do you manage sessions in Flask?
from flask import Flask, session, request, redirect, url_for

app = Flask(__name__)
app.secret_key = 'my_secret_key'

@app.route('/set_session')
def set_session():
    session['key'] = 'value'
    return 'Session set'

@app.route('/get_session')
def get_session():
    if 'key' in session:
        return session['key']
    else:
        return 'Session not set'

@app.route('/delete_session')
def delete_session():
    session.pop('key', None)
    return 'Session deleted'

@app.route('/clear_session')
def clear_session():
    session.clear()
    return 'Session cleared'
```

```python
# 9. How do you redirect to a different route in Flask?
from flask import Flask, redirect, url_for

app = Flask(__name__)

@app.route('/home')
def home():
    return 'You are on the home page'

@app.route('/admin')
def admin():
    return 'You are on the admin page'

@app.route('/go_to_admin')
def go_to_admin():
    return redirect(url_for('admin'))

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 10. How do you handle errors in Flask (e.g., 404)?

@app.errorhandler(404)
def page_not_found(e):
    return "This page does not exist", 404

@app.errorhandler(500)
def internal_server_error(e):
    return "An internal server error occurred", 500

@app.errorhandler(Exception)
def handle_exception(e):
    return "An error occurred: {}".format(e), 500
```

```python
# 11. How do you structure a Flask app using Blueprints?

from flask import Blueprint, Flask

# Create a Flask app
app = Flask(__name__)

# Create a Blueprint
my_blueprint = Blueprint('my_blueprint', __name__)

# Define a route in the Blueprint
@my_blueprint.route('/my_route')
def my_route():
    return 'Hello, World!'

# Register the Blueprint with the Flask app
app.register_blueprint(my_blueprint)

# Run the Flask app
if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 12. How do you define a custom Jinja filter in Flask?

from flask import Flask
app = Flask(__name__)

@app.template_filter('custom_filter')
def custom_filter(s):
    return s.upper()

@app.route('/')
def index():
    return render_template('index.html')

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# 15. How do you capture URL parameters in Flask?

from flask import Flask, request

app = Flask(__name__)

@app.route('/user/<username>')
def show_user_profile(username):
    # show the user profile for that user
    return 'User %s' % username

@app.route('/post/<int:post_id>')
def show_post(post_id):
    # show the post with the given id, the id is an integer
    return 'Post %d' % post_id

@app.route('/path/<path:subpath>')
def show_subpath(subpath):
    # show the subpath after /path/
    return 'Subpath %s' % subpath

@app.route('/<string:length>')
def show_length(length):
    # show the length of the string
    return 'Length %s' % length

@app.route('/<any(Admin, User):role>')
def show_role(role):
    # show the role of the user
    return 'Role %s' % role

@app.route('/<int:year>/<string:month>/<int:day>')
def show_date(year, month, day):
    # show the date
    return 'Date %d/%s/%d' % (year, month, day)

if __name__ == '__main__':
    app.run(debug=True)
```
