## Theoretical Questions

1. What is Object-Oriented Programming (OOP)?

Ans. Object-Oriented Programming (OOP) is a programming paradigm that revolves around objects and classes. It emphasizes the concept of objects, which have properties and methods that define their behavior. OOP is based on the idea of encapsulation, inheritance, and polymorphism. It allows developers to create reusable code, improve code organization, and enhance maintainability. OOP is a fundamental concept in software development and is widely used in various programming languages.

2. What is a class in OOP?

Ans. A class is a blueprint or a template that defines the properties and behavior of an object. It is a collection of data members (attributes) and member functions (methods) that define the charecteristics and actions of an object. A class is essentially a design or a template that defines the structure and behavior of an object. It is a way to define a set of attributes and methods that can be used to create objects.

3. What is an object in OOP?

Ans. An object is an instance of the class. It has its own set of attributes and methods. It is a real-world entity that can be represented in the program. It is a collection of data and methods that operate on that data. It is a runtime entity. 

4. What is the difference between abstraction and encapsulation?

Ans. Abstraction is the process of hiding the internal details of an object from the outside world, while encapsulation is the process of bundling data and methods that operate on that data within a single unit class or object. Abstraction is a higher-level concept that focuses on hiding the implementation details, while encapsulation is a lower-level concept that focuses on bundling data and methods.

5. What are dunder methods in Python?

Ans. Dunder methods are special methods in Python that enclosed in double underscore (__) before and after the method name. These methods are used for operator overloading and are called when the corresponding operator is used on an object. For example, `__add__` is called when the `+` operator is used on an object. Dunder methods are used to define how an object should behave when used with operators. They are also known as magic methods or special methods.

6. Explain the concept of inheritance in OOP.

Ans. Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows one class to inherit the properties and behavior of another class. The inheriting class is called the subclass or derived class, and the class being inherited from is called the superclass or base class. Inheritance enables code reuse, promotes modularity, and facilitates the creation of a hierarchy of related classes. It allows a subclass to inherit all the fields and methods of a superclass and can also add new fields and methods or override the ones inherited from the superclass. 

7. What is polymorphism in OOP?

Ans. Polymorphism is the ability of an object to take on multiple forms. It is a fundamental concept in object-oriented programming (OOP) that allows objects of different classes to be treated as objects of a common superclass. Polymorphism enables objects to be used in a generic way, without knowing their specific class or type. It is achieved through method overriding or method overloading. 

8. How is encapsulation achieved in Python?

Ans. Encapsulation in Python is achieved through the use of classes and objects. When you define a class, you can include variables and methods that are part of that class. These variables and methods are encapsulated within the class and can only be accessed through the class's methods. This is known as data hiding. 

9. What is a constructor in Python?

Ans. A constructor in Python is a special method that is automatically called when an object of a class is instantiated. It is used to initialize the attributes of the class. The constructor is defined with the name `__init__` and it takes `self` as the first argument. The constructor is called when an object of the class is created. It is used to set the initial state of the object. For example: `class Person: def __init__(self, name, age): self.name = name self.age = age` 

10. What are class and static methods in Python?

Ans. Class methods are methods that are bound to a class rather than an instance of the class. They are used to define methods that can be called without creating an instance of the class. Static methods are methods that are bound to a class and do not require an instance of the class to be called. They are used to define methods that do not depend on the state of the class. Both class and static methods are used to provide a way to group related functionality together and make it easier to use. They are also used to provide a way to create a singleton class, where only one instance of the class can be created.

11. What is method overloading in Python?  

Ans. Python does not support traditional method overloading. Defining multiple methods with the same name in a class will result in the last definition overriding the previous ones.

12. What is method overriding in OOP? 
 
Ans. Method overriding occurs when a subclass provides its own implementation of a method already defined in its parent class, allowing for customized behavior.

13. What is a property decorator in Python?  

Ans. A property decorator (using @property) allows a method to be accessed like an attribute, enabling controlled access and the ability to add logic to attribute retrieval.

14. Why is polymorphism important in OOP?  

Ans. Polymorphism enables objects of different classes to be treated uniformly based on shared behavior, leading to flexible and reusable code.

15. What is an abstract class in Python?  

Ans. An abstract class cannot be instantiated and serves as a blueprint for other classes. It typically contains one or more abstract methods that must be implemented by its subclasses, often using the abc module.

16. What are the advantages of OOP?  

Ans. The advantages of OOP include code reusability, easier maintenance through inheritance and encapsulation, improved code organization, and the ability to model real-world entities effectively.


17. What is the difference between a class variable and an instance variable?  

Ans. Class variables are shared among all instances of a class, while instance variables are unique to each object instance.

18. What is multiple inheritance in Python?  

Ans. Multiple inheritance allows a class to inherit attributes and methods from more than one parent class, combining behaviors from multiple sources.

19. Explain the purpose of __str__ and __repr__ methods in Python.  

Ans. The __str__ method returns a user-friendly string representation of an object, whereas the __repr__ method returns an unambiguous string representation that is useful for debugging and, ideally, recreating the object.

20. What is the significance of the super() function in Python?  

Ans. The super() function allows you to call methods from a parent class without explicitly naming it, which simplifies the code and helps manage multiple inheritance.

21. What is the significance of the __del__ method in Python?  

Ans. The __del__ method acts as a destructor that is invoked when an object is about to be garbage collected, allowing for cleanup of resources; however, its use is generally discouraged due to unpredictability.


22. What is the difference between @staticmethod and @classmethod in Python?  

Ans. A @staticmethod does not take an implicit first argument (neither self nor cls) and behaves like a regular function within a class. In contrast, a @classmethod takes the class (cls) as its first argument, which allows it to modify the class state.

23. How does polymorphism work in Python with inheritance?  

Ans. Through inheritance, subclasses can override methods from a base class. Polymorphism lets you write code that calls methods on objects without needing to know their exact type, as the appropriate overridden method is invoked.


24. What is method chaining in Python OOP?  

Ans. Method chaining is a technique where each method returns self (or another object), allowing multiple methods to be called sequentially in one statement.

25. What is the purpose of the __call__ method in Python?  

Ans. The __call__ method makes an instance callable like a function, enabling the object to execute code when it is "called" directly.

```python
# 1. Create a parent class `Animal` with a method `speak()` that prints a generic message. Create a child class `Dog` that overrides the `speak()` method to print "Bark!".

class Animal:
    def speak(self):
        print("I am an animal")

class Dog(Animal):
    def speak(self):
        print("Bark!")

dog = Dog()
dog.speak()
```

```python
# 2. Write a program to create an abstract class `Shape` with a method `area()`. Derive classes `Circle` and `Rectangle` from it and implement the `area()` method in both.

from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        return self.length * self.breadth

circle = Circle(5)
print(circle.area())
rectangle = Rectangle(5, 10)
print(rectangle.area())
```

```python
# 3. Implement a multi-level inheritance scenario where a class `Vehicle` has an attribute `type`. Derive a class `Car` and further derive a class `ElectricCar` that adds a `battery` attribute.

class Vehicle:
    def __init__(self, type):
        self.type = type
        self.wheels = 4
        self.engine = True

class Car(Vehicle):
    def __init__(self, type, brand):
        super().__init__(type)
        self.brand = brand

class ElectricCar(Car):
    def __init__(self, type, brand, battery):
        super().__init__(type, brand)
        self.battery = battery

e_car = ElectricCar("Electric", "Tesla", "100 kWh")
print(e_car.type)
print(e_car.brand)   
print(e_car.battery) 
```

```python
# 4. Demonstrate polymorphism by creating a base class `Bird` with a method `fly()`. Create two derived classes `Sparrow` and `Penguin` that override the `fly()` method.

class Bird:
    def fly(self):
        print("Bird is flying")
class Sparrow(Bird):
    def fly(self):
        print("Sparrow is flying")
class Penguin(Bird):
    def fly(self):
        print("Penguin does not fly")

# Test the classes
bird = Bird()
sparrow = Sparrow()
penguin = Penguin()
bird.fly()
sparrow.fly()
penguin.fly()
```

```python
# 5. Write a program to demonstrate encapsulation by creating a class `BankAccount` with private attributes `balance` and methods to deposit, withdraw, and check balance.

class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive")
    
    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient funds")
        elif amount > 0:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")
        else:
            print("Withdrawal amount must be positive")
    
    def get_balance(self):
        return self.__balance

# Test the BankAccount class
account = BankAccount(100)
account.deposit(50)
account.withdraw(30)
print("Final balance:", account.get_balance())
```

```python
# 6. Demonstrate runtime polymorphism using a method `play()` in a base class `Instrument`. Derive classes `Guitar` and `Piano` that implement their own version of `play()`.

class Instrument:
    def play(self):
        pass

class Guitar(Instrument):
    def play(self):
        print("Guitar is playing")

class Piano(Instrument):
    def play(self):
        print("Piano is playing")

# Demonstrate runtime polymorphism
instruments = [Guitar(), Piano()]
for instrument in instruments:
    instrument.play()
```

```python
# 7. Create a class `MathOperations` with a class method `add_numbers()` to add two numbers and a static method `subtract_numbers()` to subtract two numbers.

class MathOperations:
    @classmethod
    def add_numbers(cls, num1, num2):
        return num1 + num2
    
    @staticmethod
    def subtract_numbers(num1, num2):
        return num1 - num2

print(MathOperations.add_numbers(10, 15))      
print(MathOperations.subtract_numbers(15, 10)) 
```

```python
# 8. Implement a class `Person` with a class method to count the total number of persons created.

class Person:
    count = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.count += 1

    @classmethod
    def get_total_persons(cls):
        return cls.count

# Example usage:
p1 = Person("Alice", 30)
p2 = Person("Bob", 25)
print("Total persons created:", Person.get_total_persons())
```

```python
# 9. Write a class `Fraction` with attributes `numerator` and `denominator`. Override the `__str__` method to display the fraction as "numerator/denominator".

class Fraction:
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"

fraction = Fraction(3, 4)
print(fraction)
```

```python
# 10. Demonstrate operator overloading by creating a class `Vector` and overriding the `__add__` method to add two vectors.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(1, 2)
v2 = Vector(3, 4)
print(v1 + v2)
```

```python
# 12. Implement a class `Student` with attributes `name` and `grades`. Create a method `average_grade()` to compute the average of the grades.

class Student:
    def __init__(self, name=None, grades=None):
        self.name = name
        self.grades = grades if grades is not None else []
    
    def average_grade(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)
    
student = Student()
student.name = "John"
student.grades = [90, 80, 70]
print(student.average_grade())
```

```python
# 13. Create a class `Rectangle` with methods `set_dimensions()` to set the dimensions and `area()` to calculate the area.

class Rectangle:
    def __init__(self):
        self.length = 0
        self.breadth = 0

    def set_dimensions(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

rect = Rectangle()
rect.set_dimensions(5, 10)
print("Area:", rect.area())
```

```python
# 14. Create a class `Employee` with a method `calculate_salary()` that computes the salary based on hours worked and hourly rate. Create a derived class `Manager` that adds a bonus to the salary.

class Employee:
    def __init__(self, hours_worked, hourly_rate):
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class Manager(Employee):
    def __init__(self, hours_worked, hourly_rate, bonus):
        super().__init__(hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        base_salary = super().calculate_salary()
        return base_salary + self.bonus

emp = Employee(40, 25)
mgr = Manager(40, 25, 500)
print("Employee salary:", emp.calculate_salary())
print("Manager salary:", mgr.calculate_salary())
```

```python
# 15. Create a class `Product` with attributes `name`, `price`, and `quantity`. Implement a method `total_price()` that calculates the total price of the product.

class Product:
    def __init__(self, name = None, price = None, quantity = None):
        self.name = name
        self.price = price
        self.quantity = quantity
    def total_price(self):
        return self.price * self.quantity
    
product = Product()
product.name = "Apple"
product.price = 10
product.quantity = 5
print(product.total_price())
```

```python
# 16. Create a class `Animal` with an abstract method `sound()`. Create two derived classes `Cow` and `Sheep` that implement the `sound()` method.

from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        print("Moo!")

class Sheep(Animal):
    def sound(self):
        print("Baa!")

# Testing the classes
cow = Cow()
sheep = Sheep()
cow.sound()
sheep.sound()
```

```python
# 17. Create a class `Book` with attributes `title`, `author`, and `year_published`. Add a method `get_book_info()` that returns a formatted string with the book's details.

class Book:
    def __init__(self, title = None, author = None, year_published = None):
        self.title = title
        self.author = author
        self.year_published = year_published
    def get_book_info(self):
        return f"Title: {self.title}\nAuthor: {self.author}\nYear Published: {self.year_published}"
    
book = Book()
book.title = "1984"
book.author = "George Orwell"
book.year_published = 1949
print(book.get_book_info())
```

```python
# 18. Create a class `House` with attributes `address` and `price`. Create a derived class `Mansion` that adds an attribute `number_of_rooms`.

class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        super().__init__(address, price)
        self.number_of_rooms = number_of_rooms

mansion_obj = Mansion("742 Evergreen Terrace", 1000000, 5)
print("Address:", mansion_obj.address)
print("Price:", mansion_obj.price)
print("Number of Rooms:", mansion_obj.number_of_rooms)
```
