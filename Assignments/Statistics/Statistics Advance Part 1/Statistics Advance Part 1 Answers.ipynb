### 1. What is a random variable in probability theory?
A random variable is a numerical outcome of a random phenomenon. It assigns a real number to each possible outcome in a sample space, allowing probabilistic analysis of events.
### 2. What are the types of random variables?
Random variables are mainly of two types: discrete (which take countable values) and continuous (which take any value within a range).
### 3. What is the difference between discrete and continuous distributions?
Discrete distributions describe random variables with countable outcomes (e.g., number of heads in coin tosses), while continuous distributions describe variables with infinite possible values within an interval (e.g., height, weight).
### 4. What are probability distribution functions (PDF)?
A probability distribution function (PDF) describes the likelihood of different outcomes for a random variable. For continuous variables, the PDF gives the probability density at each value.
### 5. How do cumulative distribution functions (CDF) differ from probability distribution functions (PDF)?
The CDF gives the probability that a random variable is less than or equal to a certain value, while the PDF describes the probability density at each specific value (for continuous variables).
### 6. What is a discrete uniform distribution?
A discrete uniform distribution is one where all possible outcomes are equally likely. For example, rolling a fair die has a discrete uniform distribution over 1 to 6.
### 7. What are the key properties of a Bernoulli distribution?
A Bernoulli distribution models a random experiment with only two possible outcomes (success or failure), with a single parameter p representing the probability of success.
### 8. What is the binomial distribution, and how is it used in probability?
The binomial distribution describes the number of successes in a fixed number of independent Bernoulli trials, each with the same probability of success. It is used to model scenarios like the number of heads in multiple coin tosses.
### 9. What is the Poisson distribution and where is it applied?
The Poisson distribution models the number of times an event occurs in a fixed interval of time or space, given the events happen independently and at a constant average rate. It is used in fields like queuing theory and reliability engineering.
### 10. What is a continuous uniform distribution?
A continuous uniform distribution describes a random variable that has an equal probability of taking any value within a specified interval. The probability density is constant across the interval.
### 11. What are the characteristics of a normal distribution?
A normal distribution is symmetric, bell-shaped, and defined by its mean and standard deviation. Most values cluster around the mean, with probabilities decreasing as you move away from the center.
### 12. What is the standard normal distribution, and why is it important?
The standard normal distribution is a normal distribution with mean 0 and standard deviation 1. It is important because many statistical methods and tables are based on it, allowing for easy calculation of probabilities and Z-scores.
### 13. What is the Central Limit Theorem (CLT), and why is it critical in statistics?
The CLT states that the sampling distribution of the sample mean approaches a normal distribution as the sample size increases, regardless of the population's distribution. It is critical because it justifies the use of normal-based inference in many situations.
### 14. How does the Central Limit Theorem relate to the normal distribution?
The CLT explains why the normal distribution arises so frequently in statistics: averages of large samples from any distribution tend to be normally distributed.
### 15. What is the application of Z statistics in hypothesis testing?
Z statistics are used to test hypotheses about population means when the population standard deviation is known. They help determine whether observed data significantly differ from a hypothesized value.
### 16. How do you calculate a Z-score, and what does it represent?
A Z-score is calculated as (X - μ) / σ, where X is the value, μ is the mean, and σ is the standard deviation. It represents how many standard deviations a value is from the mean.
### 17. What are point estimates and interval estimates in statistics?
A point estimate gives a single value as an estimate of a population parameter (e.g., sample mean), while an interval estimate provides a range of values (e.g., confidence interval) likely to contain the parameter.
### 18. What is the significance of confidence intervals in statistical analysis?
Confidence intervals provide a range of plausible values for a population parameter, reflecting the uncertainty in estimation. They help quantify the reliability of estimates.
### 19. What is the relationship between a Z-score and a confidence interval?
Z-scores are used to calculate confidence intervals for population parameters when the standard deviation is known. The Z-score determines the width of the interval for a given confidence level.
### 20. How are Z-scores used to compare different distributions?
Z-scores standardize values from different distributions, allowing direct comparison by expressing values in terms of standard deviations from their respective means.
### 21. What are the assumptions for applying the Central Limit Theorem?
The main assumptions are: samples are independent, identically distributed, and the sample size is sufficiently large (usually n ≥ 30).
### 22. What is the concept of expected value in a probability distribution?
The expected value is the long-run average outcome of a random variable, calculated as the sum (or integral) of all possible values weighted by their probabilities.
### 23. How does a probability distribution relate to the expected outcome of a random variable?
A probability distribution specifies the likelihood of each possible outcome, and the expected value summarizes the average outcome based on these probabilities.
