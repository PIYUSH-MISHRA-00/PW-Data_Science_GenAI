### 1. What is hypothesis testing in statistics?
Hypothesis testing is a statistical method used to make decisions about population parameters based on sample data. It involves formulating a null hypothesis and an alternative hypothesis, then using sample data to determine which hypothesis is more likely.
### 2. What is the null hypothesis, and how does it differ from the alternative hypothesis?
The null hypothesis (H0) is a statement that there is no effect or no difference, and it serves as the default assumption. The alternative hypothesis (H1) is what you want to prove, indicating the presence of an effect or difference. Hypothesis testing evaluates which hypothesis is supported by the data.
### 3. Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.
The significance level (alpha, usually 0.05) is the probability of rejecting the null hypothesis when it is actually true (Type I error). It sets the threshold for statistical evidence required to declare a result significant.
### 4. What are Type I and Type II errors? Give examples of each.
Type I error occurs when the null hypothesis is wrongly rejected (false positive). Example: Concluding a drug works when it does not. Type II error occurs when the null hypothesis is wrongly accepted (false negative). Example: Failing to detect a real effect of a drug.
### 5. What is the difference between a Z-test and a T-test? Explain when to use each.
A Z-test is used when the population standard deviation is known and the sample size is large (n > 30). A T-test is used when the population standard deviation is unknown and/or the sample size is small (n ≤ 30).
### 6. Write a Python program to generate a binomial distribution with n=10 and p=0.5, then plot its histogram.
```python
import numpy as np
import matplotlib.pyplot as plt
n = 10
p = 0.5
size = 1000
binom_data = np.random.binomial(n, p, size)
plt.hist(binom_data, bins=range(n+2), edgecolor='black', alpha=0.7)
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of Successes')
plt.ylabel('Frequency')
plt.show()
```
### 7. Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.
```python
import numpy as np
from scipy.stats import norm
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]
mu = 50 # Hypothesized population mean
sigma = 1 # Population standard deviation (assumed known)
n = len(sample_data)
mean_sample = np.mean(sample_data)
z = (mean_sample - mu) / (sigma / np.sqrt(n))
p_value = 2 * (1 - norm.cdf(abs(z)))
print(f"Z-statistic: {z:.2f}")
print(f"p-value: {p_value:.4f}")
if p_value < 0.05:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```
Interpretation: If the p-value is less than 0.05, there is significant evidence to reject the null hypothesis; otherwise, we fail to reject it.
### 8. Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.
```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
np.random.seed(0)
data = np.random.normal(loc=100, scale=15, size=200)
mean = np.mean(data)
std_err = np.std(data, ddof=1) / np.sqrt(len(data))
conf_int = norm.interval(0.95, loc=mean, scale=std_err)
plt.hist(data, bins=20, edgecolor='black', alpha=0.7)
plt.title('Simulated Normal Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
print(f"Mean: {mean:.2f}")
print(f"95% Confidence Interval: {conf_int}")
```
### 9. Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.
```python
import numpy as np
import matplotlib.pyplot as plt
def calculate_z_scores(data):
    mean = np.mean(data)
    std = np.std(data)
    return [(x - mean) / std for x in data]
data = np.random.normal(50, 5, 100)
z_scores = calculate_z_scores(data)
plt.hist(z_scores, bins=20, edgecolor='black', alpha=0.7)
plt.title('Histogram of Z-scores')
plt.xlabel('Z-score')
plt.ylabel('Frequency')
plt.show()
```
Z-scores indicate how many standard deviations a data point is from the mean. A Z-score of 0 means the value is at the mean, positive values are above the mean, and negative values are below the mean.
