## Theoretical Questions

1. What is statistics, and why is it important?

Ans. Statistics is the study of the collection, analysis, interpretation, presentation, and organization of data. It is important because it provides tools for making informed decisions based on data, helps in understanding trends and patterns, and supports scientific research by allowing for the testing of hypotheses and theories. It is widely used in various fields such as economics, medicine, social sciences, and business to analyze data and draw conclusions that can influence policy and practice.

2. What are the two main types of statistics?

Ans. The two main types of statistics are descriptive statistics and inferential statistics.

3. What are descriptive statistics?

Ans. Descriptive statistics involve methods for summarizing and organizing data, such as calculating means, medians, modes, ranges, and creating visual representations like histograms or boxplots.

4. What is inferential statistics?

Ans. Inferential statistics use sample data to make generalizations or predictions about a larger population. They involve estimation, hypothesis testing, and drawing conclusions based on probabilistic models.

5. What is sampling in statistics?

Ans. Sampling in statistics is the process of selecting a subset of individuals or observations from a larger population to estimate characteristics of that population.

6. What are the different types of sampling methods?

Ans. Common sampling methods include simple random sampling, stratified sampling, cluster sampling, systematic sampling, and convenience sampling, among others.

7. What is the difference between random and non-random sampling?

Ans. Random sampling gives each member of the population an equal chance of being selected, reducing bias, while non-random sampling does not, which can introduce bias into the sample.

8. Define and give examples of qualitative and quantitative data.

Ans. Qualitative data describes qualities or characteristics (e.g., color, type, or brand), while quantitative data consists of numerical values (e.g., age, height, or income) that can be measured and analyzed statistically.

9. What are the different types of data in statistics?

Ans. Data in statistics can be classified into nominal, ordinal, interval, and ratio types.

10. Explain nominal, ordinal, interval, and ratio levels of measurement.

Ans. Nominal data categorize without a specific order (e.g., gender or nationality); ordinal data provide a ranked order (e.g., class rankings); interval data have ordered, equal intervals without an absolute zero (e.g., temperature in Celsius); and ratio data have a meaningful zero, allowing for the comparison of absolute magnitudes (e.g., weight or height).

11. What is the measure of central tendency?

Ans. Measures of central tendency are statistical metrics that describe the center or typical value of a dataset, such as the mean, median, and mode.

12. Define mean, median, and mode.

Ans. The mean is the average of a dataset, the median is the middle value when the data is ordered, and the mode is the most frequently occurring value in the dataset.

13. What is the significance of the measure of central tendency?

Ans. Measures of central tendency summarize a dataset with a single value that represents its center, aiding in the understanding and comparison of data distributions.

14. What is variance, and how is it calculated?

Ans. Variance measures the spread of a set of data by calculating the average squared deviation of each data point from the mean.

15. What is standard deviation, and why is it important?

Ans. Standard deviation is the square root of the variance, providing a measure of dispersion in the same units as the data. It is important because it indicates how much the data deviates from the mean on average.

16. Define and explain the term range in statistics.

Ans. The range is the difference between the maximum and minimum values in a dataset, serving as a basic measure of variability.

17. What is the difference between variance and standard deviation?

Ans. Variance is the average of the squared differences from the mean, while standard deviation is the square root of the variance, making it more interpretable by expressing dispersion in the original units of the data.

18. What is skewness in a dataset?

Ans. Skewness quantifies the degree of asymmetry in a dataset’s distribution. A skewed dataset has a tail that extends more to one side than the other.

19. What does it mean if a dataset is positively or negatively skewed?

Ans. If a dataset is positively skewed, the right tail is longer, indicating that a majority of the data is concentrated on the left. Conversely, a negatively skewed dataset has a longer left tail, meaning most data points are on the right.

20. Define and explain kurtosis.

Ans. Kurtosis measures the “tailedness” of a data distribution, indicating whether data are heavy-tailed (more outliers) or light-tailed compared to a normal distribution.

21. What is the purpose of covariance?

Ans. Covariance measures the directional relationship between two variables, showing whether they tend to increase or decrease together.

22. What does correlation measure in statistics?

Ans. Correlation measures the strength and direction of the linear relationship between two variables, typically standardized to range between -1 and 1.

23. What is the difference between covariance and correlation?

Ans. Covariance provides an unstandardized measure of how two variables change together, while correlation standardizes this measure, making it easier to interpret the strength and direction of the relationship.

24. What are some real-world applications of statistics?

Ans. Statistics is applied in various fields such as economics for forecasting, healthcare for clinical trials, social sciences for behavioral research, marketing for consumer analysis, and many other domains to drive decision-making and policy formulation.

```python
# 1. How do you calculate the mean, median, and mode of a dataset?
import statistics

data = [10, 20, 20, 30, 40]  # sample dataset
mean_value = statistics.mean(data)
median_value = statistics.median(data)
mode_value = statistics.mode(data)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

```python
# 2. Write a Python program to compute the variance and standard deviation of a dataset.
variance_value = statistics.variance(data)
stdev_value = statistics.stdev(data)
print("\nVariance:", variance_value)
print("Standard Deviation:", stdev_value)
```

```python
# 4. Implement sampling techniques like random sampling and stratified sampling.
import random

# Simple random sampling from a population of numbers from 1 to 100.
population = list(range(1, 101))
random_sample = random.sample(population, 10)
print("\nRandom Sample:", random_sample)

# Stratified sampling example: split population into even and odd numbers.
stratum_even = [x for x in population if x % 2 == 0]
stratum_odd = [x for x in population if x % 2 != 0]
sample_even = random.sample(stratum_even, 5)
sample_odd = random.sample(stratum_odd, 5)
print("Stratified Sample (Even):", sample_even)
print("Stratified Sample (Odd):", sample_odd)
```

```python
# 5. Write a Python function to calculate the range of a dataset.
def calculate_range(dataset):
    return max(dataset) - min(dataset)
print("\nRange of data:", calculate_range(data))
```

```python
# 6. Create a dataset and plot its histogram to visualize skewness.
import matplotlib.pyplot as plt
import numpy as np

data_hist = np.random.randn(1000)  # normally distributed dataset
plt.figure(figsize=(6,4))
plt.hist(data_hist, bins=30, edgecolor='black')
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# 7. Calculate skewness and kurtosis of a dataset using Python libraries.
from scipy.stats import skew, kurtosis

data_stats = np.random.randn(1000)
skewness = skew(data_stats)
kurt = kurtosis(data_stats)
print("\nSkewness:", skewness)
print("Kurtosis:", kurt)
```

```python
# 8. Generate a dataset and demonstrate positive and negative skewness.
# Positive skew: exponential distribution; Negative skew: negated exponential.
data_positive_skew = np.random.exponential(scale=1, size=1000)
data_negative_skew = -np.random.exponential(scale=1, size=1000)
print("\nPositive Skewness:", skew(data_positive_skew))
print("Negative Skewness:", skew(data_negative_skew))
```

```python
# 9. Write a Python script to calculate covariance between two datasets.
x = np.random.randn(100)
y = np.random.randn(100)
covariance_matrix = np.cov(x, y)
print("\nCovariance matrix:\n", covariance_matrix)
```

```python
# 10. Write a Python script to calculate the correlation coefficient between two datasets.
correlation_matrix = np.corrcoef(x, y)
print("\nCorrelation matrix:\n", correlation_matrix)
```

```python
# 13. Calculate the mean, median, and mode of grouped data.
# Example of grouped data represented by group midpoints and frequencies.
group_midpoints = [5, 15, 25, 35]
frequencies = [10, 15, 5, 20]
weighted_mean = sum(m * f for m, f in zip(group_midpoints, frequencies)) / sum(frequencies)
print("\nWeighted Mean (Grouped Data):", weighted_mean)
# For median and mode, grouped data requires more detailed computation.
```

```python
# 14. Simulate data using Python and calculate its central tendency and dispersion.
data_simulated = np.random.normal(loc=50, scale=10, size=1000)
print("\nSimulated Data Mean:", np.mean(data_simulated))
print("Simulated Data Median:", np.median(data_simulated))
print("Simulated Data Standard Deviation:", np.std(data_simulated))
```

```python
# 15. Use NumPy or pandas to summarize a dataset’s descriptive statistics.
import pandas as pd

df_simulated = pd.DataFrame({"Values": data_simulated})
print("\nDescriptive Statistics:\n", df_simulated.describe())
```

```python
# 16. Plot a boxplot to understand the spread and identify outliers.
plt.figure(figsize=(6,4))
plt.boxplot(data_simulated, vert=False)
plt.title("Boxplot")
plt.xlabel("Value")
plt.show()
```

```python
# 17. Calculate the interquartile range (IQR) of a dataset.
q75, q25 = np.percentile(data_simulated, [75, 25])
iqr = q75 - q25
print("\nInterquartile Range (IQR):", iqr)
```

```python
# 18. Implement Z-score normalization and explain its significance.
def z_score_normalization(dataset):
    return (dataset - np.mean(dataset)) / np.std(dataset)
normalized_data = z_score_normalization(data_simulated)
print("\nFirst 5 normalized values:", normalized_data[:5])
# Z-score normalization standardizes data to have mean 0 and standard deviation 1.
```

```python
# 21. Use seaborn to create a correlation matrix for a dataset.
corr_matrix = np.corrcoef(data_matrix, rowvar=False)
plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```
